## P1. Operaciones y despeje con NumPy (15 pts)

Con **A** = [[2, −4, 6], [8, 0, −2]] y **B** = [[1, 3, −1], [2, −2, 4]] (filas entre corchetes), resuelva en el cuaderno:

**(a)** Cree A y B como arreglos de NumPy y muestre su tamaño y la transpuesta de A.

```python
A = np.array([[2, -4,  6],
              [8,  0, -2]])
B = np.array([[1,  3, -1],
              [2, -2,  4]])

print("tamaño de A:", A.shape)
print("A.T =\n", A.T)
print("tamaño de A.T:", A.T.shape)
```

```python
# VERIFICACIÓN (a)
assert A.shape == (2, 3) and B.shape == (2, 3), "A y B deben ser 2 x 3"
assert A[1, 2] == -2 and B[0, 1] == 3, "revise las entradas de A y B"
print("(a) correcto")
```

**(b)** Calcule 3A − 2B y verifique en código que $(3A-2B)^T = 3A^T - 2B^T$.

```python
C = 3*A - 2*B

print("3A =\n", 3*A)
print("2B =\n", 2*B)
print("C = 3A - 2B =\n", C)
print("\n(3A - 2B)^T =\n", C.T)
print("3A^T - 2B^T =\n", 3*A.T - 2*B.T)
print("¿Son iguales?", np.array_equal(C.T, 3*A.T - 2*B.T))
```

```python
# VERIFICACIÓN (b)
assert C.shape == (2, 3), "C debe ser 2 x 3"
assert np.array_equal(C.T, 3*A.T - 2*B.T)
print("(b) correcto")
```

**(c)** Despeje X en $4X + A = 2B$, calcule X y verifique con `np.allclose`.

```python
# Despeje paso a paso (escríbalo aquí como comentario):
#   4X + A = 2B
#   4X = 2B - A          (se suma el opuesto de A a ambos lados)
#   X = (1/4)(2B - A)    (no se divide entre una matriz: se multiplica por el inverso del escalar, 1/4)

X = 0.25 * (2*B - A)

print("X =\n", X)
print("¿4X + A == 2B ?", np.allclose(4*X + A, 2*B))
```

```python
# VERIFICACIÓN (c)
assert X.shape == (2, 3), "X debe ser 2 x 3"
assert np.allclose(4*X + A, 2*B), "X no cumple 4X + A = 2B"
print("(c) correcto")
```

**(d)** Calcule A @ B$^T$ y B$^T$ @ A. Muestre sus tamaños y explique con un comentario por qué son diferentes.

```python
P1 = A @ B.T      # (2x3)(3x2) = 2x2
P2 = B.T @ A      # (3x2)(2x3) = 3x3

print("A @ B.T =\n", P1)
print("tamaño:", P1.shape)
print("\nB.T @ A =\n", P2)
print("tamaño:", P2.shape)

# Explicación (comentario):
# El producto de matrices no es conmutativo. En A @ B.T los números de adentro son 3 y 3 y los de afuera
# 2 y 2, así que el resultado es 2x2 (cada entrada es una fila de A punto una fila de B).
# En B.T @ A los de adentro son 2 y 2 y los de afuera 3 y 3, así que el resultado es 3x3.
# Al cambiar el orden cambian el tamaño y los valores: ni siquiera pueden compararse.
```

```python
# VERIFICACIÓN (d)
assert P1.shape == (2, 2) and P2.shape == (3, 3)
print("(d) correcto")
```

## P2. Sistema de recomendación con matrices (20 pts)

Un sistema de recomendación describe a cada usuario con 2 gustos (matriz **U** de 4 × 2) y a cada película con 2 rasgos (matriz **F** de 3 × 2). Su calificación predicha es $\hat{\mathbf{R}} = \mathbf{U} \cdot \mathbf{F}^T$. En el cuaderno se dan U, F y las calificaciones reales **R** (4 × 3). Resuelva:

```python
U = np.array([[1.0, 0.5],
              [0.2, 1.0],
              [1.0, 1.0],
              [0.5, 0.2]])      # usuarios x gustos     (4 x 2)
F = np.array([[2.0, 1.0],
              [1.0, 2.0],
              [3.0, 0.5]])      # películas x rasgos    (3 x 2)
R = np.array([[4, 3, 5],
              [2, 3, 2],
              [5, 5, 4],
              [3, 2, 4]], dtype=float)   # calificaciones reales (4 x 3)
print("U", U.shape, "| F", F.shape, "| R", R.shape)
```

**(a)** Calcule $\hat R$ = U @ F.T y verifique su tamaño (4, 3).

```python
R_hat = U @ F.T      # (4x2)(2x3) = 4x3

print("R_hat =\n", R_hat)
print("tamaño:", R_hat.shape)
```

```python
# VERIFICACIÓN (a)
assert R_hat.shape == (4, 3), "R_hat debe ser 4 x 3"
assert np.isclose(R_hat[0, 0], 2.5), "revise el producto U @ F.T"
print("(a) correcto")
```

**(b)** Calcule la matriz de error E = R − $\hat R$ y el error cuadrático medio ECM = promedio de los cuadrados de E.

```python
E = R - R_hat                 # error: real menos predicho
ecm0 = np.mean(E**2)          # promedio de los cuadrados de las entradas de E

print("E =\n", E)
print("ECM inicial (ecm0):", round(ecm0, 4))
```

```python
# VERIFICACIÓN (b)
assert E.shape == (4, 3)
assert np.isclose(ecm0, np.mean((R - U @ F.T)**2))
print("(b) correcto")
```

**(c)** Actualice U con un paso de aprendizaje: U_nueva = U + η · (E @ F), con η = 0,05. Indique el tamaño de E @ F.

```python
eta = 0.05

# Tamaño de E @ F (comentario): (4x3)(3x2) = 4x2, el mismo tamaño que U, por eso se puede sumar a U.

U_nueva = U + eta * (E @ F)

print("tamaño de E @ F:", (E @ F).shape)
print("U_nueva =\n", U_nueva)
```

```python
# VERIFICACIÓN (c)
assert U_nueva.shape == (4, 2), "U_nueva debe ser 4 x 2"
print("(c) correcto")
```

**(d)** Recalcule $\hat R$ y el ECM con U_nueva. Compruebe en código que el error bajó.

```python
R_hat1 = U_nueva @ F.T
ecm1 = np.mean((R - R_hat1)**2)

print("R_hat1 =\n", R_hat1)
print("ECM antes  (ecm0):", round(ecm0, 4))
print("ECM después (ecm1):", round(ecm1, 4))
print("¿El error bajó?", ecm1 < ecm0)
```

```python
# VERIFICACIÓN (d)
assert ecm1 < ecm0, "el error debería bajar con un paso de aprendizaje"
print("(d) correcto")
```

**(e)** Con la predicción original $\hat R$, ¿qué película (columna) se recomendaría primero al usuario 0? Use `np.argmax` y escriba la respuesta con una frase.

```python
mejor = int(np.argmax(R_hat[0]))   # R_hat[0] = predicciones del usuario 0 para las 3 películas

print("Predicciones del usuario 0:", R_hat[0])
print("Película recomendada al usuario 0 (índice):", mejor)
```

```python
# VERIFICACIÓN (e)
assert mejor in (0, 1, 2)
print("(e) correcto")
```

Al usuario 0 se le recomendaría la película de índice 2 (la tercera), porque tiene la mayor calificación predicha (3,25 frente a 2,50 y 2,00). La predicción salió del producto de matrices $\hat R = U\,F^T$ (cada entrada es el producto punto entre los gustos de un usuario y los rasgos de una película). El modelo aprendió con la actualización $U_{\text{nueva}} = U + \eta\,(E\,F)$: el error $E = R - \hat R$ multiplicado por $F$ indica en qué dirección mover los gustos, y con un solo paso el ECM bajó de 2,215 a 0,263.
